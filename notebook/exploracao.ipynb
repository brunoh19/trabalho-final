```python
from utils import carregar_dados

dados = carregar_dados()
for nome, df in dados.items():
    print(nome, df.shape)
```

```python
from utils import inspecionar
for nome, df in dados.items():
    df_inspecao = df.copy()
    for coluna in df_inspecao.select_dtypes(include=["object"]).columns:
        df_inspecao[coluna] = df_inspecao[coluna].map(
            lambda valor: repr(valor) if isinstance(valor, (list, dict, set)) else valor
        )
    inspecionar(df_inspecao, nome)
```

```python
import json
hoteis = dados["hotels"]
hoteis["amenities"] = hoteis["amenities"].apply(lambda x: json.loads(x) if isinstance(x, str) else x)
comodidades = hoteis[["hotel_id", "amenities"]].explode("amenities")
comodidades.groupby("amenities")["hotel_id"].nunique().sort_values(ascending=False)
```

```python
ids = set(dados["hotels"]["hotel_id"])
for nome in ["room_types", "daily_market", "weekly_performance", "rate_decisions", "reviews"]:
    print(nome, set(dados[nome]["hotel_id"]) <= ids)
print("scenario_id", set(dados["reviews"]["scenario_id"]) <= set(dados["review_scenarios"]["scenario_id"]))
```
